```python
class Person:
    def __init__(self):
        self.__name = "John Doe"  # private attribute
        self.__age = 30           # private attribute
    
    def hello(self):
        print(f"Hello, I am a person. My name is {self.__name} and I am {self.__age} years old.")

john = Person()
john.hello()
```

```python
class Student(Person):
    # override phương thức hello() từ lớp cha Person
    def hello(self):
        print(f"Hello, I am a student. My name is {self.__name} and I am {self.__age} years old.")

paul = Student()
paul.hello()  # Gọi phương thức hello() từ lớp Student, nhưng không truy cập được name & age (private)
```

```python
class Student(Person):
    # override phương thức hello() từ lớp cha Person
    def hello(self):
        super().hello()  # Gọi phương thức hello() từ lớp cha Person nên truy cập gián tiếp được name & age
        print(f"I am a student.")

paul = Student()
paul.hello() 
```

```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age
    
    @property
    def name(self):
        return self.__name
    
    @name.setter
    def name(self, value):
        self.__name = value # skip validation for simplicity

    @property
    def age(self):
        return self.__age
    
    @age.setter
    def age(self, value):
        self.__age = value # skip validation for simplicity

    def greet(self):
        print(f"Hello, I am a person. My name is {self.name} and I am {self.age} years old.")
```

```python
class Student(Person):
    def __init__(self, name, age, sid):
        super().__init__(name, age) # Gọi hàm khởi tạo của lớp cha Person để khởi tạo name & age
        self.__student_id = sid
    
    @property
    def student_id(self):
        return self.__student_id
    
    @student_id.setter
    def student_id(self, value):
        self.__student_id = value # skip validation for simplicity

    # override (ghi đè) phương thức greet() từ lớp cha Person, sử dụng public getter để truy cập name & age
    def greet(self):
        print(f'Hello, I am a student. My name is {self.name}, I am {self.age} years old, and my student ID is {self.student_id}.')

paul = Student("Paul Smith", 20, "S12345")
paul.greet()  # Gọi phương thức greet() từ lớp Student, print name, age và student_id
```
