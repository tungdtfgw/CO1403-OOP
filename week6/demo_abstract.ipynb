```python
# import the necessary modules for abstract class and abstract method
from abc import abstractmethod, ABC

# inherit from the ABC class to create an abstract class
class MyAbstract(ABC):
    # define an abstract method that must be implemented by any subclass
    @abstractmethod
    def my_method(self):
        pass

class MyConcrete(MyAbstract):
    def my_method(self):
        print("Hello from MyConcrete!")
```

```python
a = MyConcrete()
a.my_method()  # Output: Hello from MyConcrete!
```

```python
# Cannot create object of abstract class
a = MyAbstract()
```

```python
class AnotherChild(MyAbstract):
    def hi(self):
        print("Hello from AnotherChild!")
```

```python
a = AnotherChild()
a.hi()  # Output: Hello from MyConcrete!
```
