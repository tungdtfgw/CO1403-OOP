```python
class Student:
    def __init__(self, name, age, grade):
        self.name = name
        self.age = age
        self.grade = grade

    def show(self):
        print(f'Student: {self.name}, Age: {self.age}, Grade: {self.grade}')
```

```python
john = Student('John Doe', 20, 'Merit')
john.name = 'John Smith'  # Change name
john.age = 21  # Change age
john.grade = 'Distinction'  # Change grade
john.show()
```

```python
# public attributes allow direct access and modification
# outside can set invalid values without any checks
john.name = ''
john.age = -5
john.grade = 'ABC'
john.show()
```

```python
class Student:
    def __init__(self, name, age, grade):
        self.__name = name      # Private attribute
        self.__age = age    # Private attribute
        self.grade = grade  # Public attribute

    def show(self):
        print(f'Student: {self.__name}, Age: {self.__age}, Grade: {self.grade}')
```

```python
john = Student('John Doe', 20, 'A')
print(john.grade)
```

```python
print(john.__name)  # This will raise an AttributeError
```

```python
john.show()
```

```python
class Student:
    def __init__(self, name, age, grade):
        self.__name = name
        self.__age = age
        self.__grade = grade

    def show(self):
        print(f'Student: {self.__name}, Age: {self.__age}, Grade: {self.__grade}')

    def get_name(self):
        return self.__name
    
    def set_name(self, name):
        if name == '':
            print('Name cannot be empty.')
        else:
            self.__name = name
    
    def get_age(self):
        return self.__age
    
    def set_age(self, age):
        if age < 0:
            print('Age cannot be negative.')
        else:
            self.__age = age

    def get_grade(self):
        return self.__grade

    def set_grade(self, grade):
        if grade.lower() not in ['failed', 'pass', 'merit', 'distinction']:
            print('Invalid grade.')
        else:
            self.__grade = grade.capitalize()
```

```python
john = Student('John Doe', 20, 'Merit')
print(john.__name)
```

```python
print(john.get_name())
```

```python
john.set_name('John Smith')
john.show()
```

```python
john.set_name('')
john.set_age(-5)
john.set_grade('ABC')
john.show()
```
