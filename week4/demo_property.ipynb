```python
class Student:
    def __init__(self, name, age):
        self.__name = name # private attribute
        self.__age = age   # private attribute
    
    # Use @property decorator to create a getter for the name attribute
    @property
    def name(self):
        return self.__name

    # Use @name.setter to create a setter for the name attribute
    @name.setter
    def name(self, new_value):
        if new_value == '':
            print("Name cannot be empty.")
        else:
            self.__name = new_value

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self, new_value):
        if new_value < 0:
            print("Age cannot be negative.")
        else:
            self.__age = new_value
```

```python
john = Student("John Doe", 20)
print(john.name)  # Calling the getter for name, not public attribute
john.name = "John Smith"  # Calling the setter for name, not public attribute
print(john.name)  
```

```python
john.name = ''
john.age = -5
```

```python
print(john.name, john.age) # still old values, not updated due to validation in setters
```

```python
class Rectangle:
    def __init__(self, width, height):
        self.__width = width
        self.__height = height

    @property
    def width(self):
        return self.__width

    @width.setter
    def width(self, new_value):
        if new_value <= 0:
            print("Width must be positive.")
        else:
            self.__width = new_value

    @property
    def height(self):
        return self.__height

    @height.setter
    def height(self, new_value):
        if new_value <= 0:
            print("Height must be positive.")
        else:
            self.__height = new_value

    @property
    def area(self): # property area is read-only, no setter
        return self.__width * self.__height
    
    def show(self):
        print(f'Rectangle {self.__width} x {self.__height}, area: {self.area}')
```

```python
abcd = Rectangle(4, 5)
abcd.show()

abcd.width = 10
abcd.height = 20
print(abcd.area) # area is updated automatically when width and height are changed
```
