```python
a = ['apple', 'banana', 'kiwi']
i = int(input('Enter an index: '))
fruit = a[i]    # IndexError if i is out of range ([-n,..,-1,0,..,(n-1)])
print(fruit)
```

```python
# without exception, program will crash when runtime error happens
a = int(input('Enter a: '))
print(a)
print('End')
```

```python
# with exception, error will be handled carefully
try:
    a = int(input('Enter a: '))
    print(a)
except ValueError as e:
    print(e)

print('End')
```

```python
try:
    a = int(input('Enter a: '))
    b = int(input('Enter b: '))
    c = a / b   # ZeroDivisionError if b = 0

    print(f'{a} / {b} = {c}')
except ZeroDivisionError as zeror_er:
    #print(zeror_er) # print default error message
    print('Không được chia cho số 0') # print custom error message
except ValueError:
    print('Phải nhập vào số nguyên')
```

```python
try:
    a = ['apple', 'banana', 'kiwi']
    i = int(input('Enter an index: '))
    fruit = a[i]    # IndexError if i is out of range ([-n,..,-1,0,..,(n-1)])
    print(fruit)
except IndexError:
    print('Index nằm ngoài khoảng cho phép')
except ValueError:
    print('Phải nhập vào số nguyên')
```

```python
# Phát hiện các errors có thể xảy ra và dùng ngoại lệ để xử lý
try:
    a = [0, 1, 2, 3, 4, 5]
    i = int(input('Enter an index: '))
    b = 1/a[i]
    print(f'1/{a[i]} = {b}')
except ValueError:
    print('Phải nhập số nguyên')
except IndexError:
    print('Index phải nằm trong khoảng cho phép')
except ZeroDivisionError:
    print('Không chia được cho 0')
except Exception:
    print('Unknown error')
```
