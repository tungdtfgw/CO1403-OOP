```python
# Define a class
class LightBulb:
    # define a constructor: __init__(self, parameter1, parameter2, ...)
    def __init__(self, room, color):
        # syntax: self.attribute = parameter
        self.room = room
        self.color = color
        self.status = 'off'
    
    # define a method: method_name(self, parameter1, parameter2, ...)
    def turn_on(self):
        self.status = 'on'

    def turn_off(self):
        self.status = 'off'

    def show(self):
        print(f'{self.color} light bulb in {self.room} is {self.status}.')
```

```python
# create object: object_name = ClassName(parameter1, parameter2, ...) => it will call __init__(self, parameter1, parameter2, ...)
living_room_light = LightBulb('living room', 'white')
bedroom_light = LightBulb('bedroom', 'yellow')
# call method: object_name.method_name(parameter1, parameter2, ...)
living_room_light.show()
bedroom_light.show()
```

```python
living_room_light.turn_on()
living_room_light.show()

living_room_light.turn_off()
living_room_light.show()
```

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def show(self):
        print(f'Product {self.name}, price {self.price}.')

    def increase(self, amount):
        self.price += amount
        print(f'Price updated. New price: {self.price}.')

    def decrease(self, amount):
        self.price -= amount
        print(f'Price updated. New price: {self.price}.')
```

```python
laptop = Product('Laptop', 1000)
laptop.show()
laptop.increase(100)
laptop.show()
laptop.decrease(50)
laptop.show()
```

```python
class Store:
    def __init__(self, name):
        # init an empty list to store products
        self.products = []
        self.name = name
    
    def add(self, p):
        self.products.append(p)
        print(f'Product {p.name} added to store {self.name}.')

    def show(self):
        print(f'All products in store {self.name}:')
        for p in self.products:
            p.show()
```

```python
# create some objects of Product class
laptop = Product('Laptop', 1000)
monitor = Product('Monitor', 200)
keyboard = Product('Keyboard', 50)
mouse = Product('Mouse', 2)
# create an object of Store class
store = Store('Tech Store')
# add products to store
store.add(laptop)
store.add(monitor)
store.add(keyboard)
store.add(mouse)
# call method to show all products in store
store.show()
```
