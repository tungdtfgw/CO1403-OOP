```python
# Khai báo 1 từ điển gồm các cặp key-value:
en_vn = {
    'hello': 'xin chào',        # hello là key, xin chào là value
    'goodbye': 'tạm biệt',      # goodbye là key, tạm biệt là value
    'thank you': 'cảm ơn',      # thank you là key, cảm ơn là value
}
```

```python
# truy cập value thông qua key:
print(en_vn['hello'])        # in ra: xin chào
print(en_vn['goodbye'])      # in ra: tạm biệt
print(en_vn['thank you'])    # in ra: cảm ơn
```

```python
# Thêm 1 cặp key-value mới vào từ điển:
en_vn['cat'] = 'mèo'
en_vn['dog'] = 'chó'

# In ra toàn bộ từ điển:
print(en_vn)
```

```python
# Sửa value theo key
en_vn['goodbye'] = 'lượn!'
print(en_vn)
```

```python
# xóa 1 cặp key-value theo key:
del en_vn['thank you']
print(en_vn)
```
